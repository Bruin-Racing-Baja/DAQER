```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np 

df = pd.read_csv("log_2025-11-15_12-46-31.csv")

df["TimeSec"] = (df["Timestamp"] - df["Timestamp"].iloc[0]) / 1e6


dt = np.diff(df["Timestamp"]) / 1e6

freq = 1 / dt

t = df["Timestamp"].iloc[1:] / 1e6

print("Average frequency:", np.mean(freq), "Hz")
print("Min frequency:", np.min(freq), "Hz")
print("Max frequency:", np.max(freq), "Hz")


plt.figure(figsize=(10,6))
plt.plot(df["TimeSec"], df["ax"], label="ax")
plt.plot(df["TimeSec"], df["ay"], label="ay")
plt.plot(df["TimeSec"], df["az"], label="az")

plt.xlabel("Time (s)")
plt.ylabel("Acceleration (m/s²)")
plt.title("IMU Accelerations")
plt.legend()
plt.grid(True)
plt.show()
```
